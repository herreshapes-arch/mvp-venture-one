Convertimos los tres archivos de Excel en tablas dentro de una base de datos SQL.

```python

consulta = """
SELECT
    i.id_inmueble,
    i.id_anuncio,
    i.id_ubicacion,

    -- Datos del anuncio
    a.fuente,
    a.zona AS zona_anuncio,
    a.municipio AS municipio_anuncio,
    a.barrio AS barrio_anuncio,
    a.precio,
    a.habitaciones,
    a.banos,
    a.parqueadero,
    a.area_m2 AS area_anuncio,

    -- Datos del inmueble
    i.origen AS origen_inmueble,
    i.direccion,
    i.area_m2 AS area_inmueble,
    i.estado AS estado_inmueble,
    i.lat,
    i.lon,

    -- Datos de ubicación
    u.zona AS zona_ubicacion,
    u.municipio AS municipio_ubicacion,
    u.barrio AS barrio_ubicacion

FROM Inmuebles_prueba AS i

LEFT JOIN Anuncios_muestra AS a
    ON i.id_anuncio = a.id_anuncio

LEFT JOIN ubicaciones AS u
    ON i.id_ubicacion = u.id_ubicacion;
"""

resultado = pd.read_sql_query(consulta, conexion)

display(resultado.head(10))
```

```python

import pandas as pd
import numpy as np

# Cargar la base consolidada
df = pd.read_excel("base_inmobiliaria_consolidada.xlsx")

# Crear una copia para conservar los datos originales
df_limpia = df.copy()

print("Registros iniciales:", len(df_limpia))
print("Columnas:", df_limpia.columns.tolist())
```

```python

# Convertir los valores no numéricos en valores nulos
df_limpia["parqueadero"] = pd.to_numeric(
    df_limpia["parqueadero"].replace(
        ["-", "", " ", "N/A"],
        np.nan
    ),
    errors="coerce"
)

# Los valores faltantes se interpretan como cero
df_limpia["parqueadero"] = (
    df_limpia["parqueadero"]
    .fillna(0)
    .astype(int)
)

print("Valores de parqueadero corregidos:")
display(
    df_limpia[
        ["id_inmueble", "parqueadero"]
    ]
)
```

```python

def normalizar_coordenada(valor, tipo):

    if pd.isna(valor):
        return np.nan

    numero = float(valor)

    # Si ya está en formato decimal, conservarlo
    if tipo == "lat" and -90 <= numero <= 90:
        return numero

    if tipo == "lon" and -180 <= numero <= 180:
        return numero

    # Separar signo y parte numérica
    signo = -1 if numero < 0 else 1
    digitos = str(int(abs(numero)))

    # Latitud: un dígito antes del decimal
    if tipo == "lat":
        numero_normalizado = int(digitos) / (10 ** (len(digitos) - 1))

    # Longitud: dos dígitos antes del decimal
    elif tipo == "lon":
        numero_normalizado = int(digitos) / (10 ** (len(digitos) - 2))

    else:
        raise ValueError("Tipo debe ser lat o lon")

    return signo * numero_normalizado


# Aplicar la normalización
df_limpia["lat_original"] = df_limpia["lat"]
df_limpia["lon_original"] = df_limpia["lon"]

df_limpia["lat"] = df_limpia["lat"].apply(
    lambda x: normalizar_coordenada(x, "lat")
)

df_limpia["lon"] = df_limpia["lon"].apply(
    lambda x: normalizar_coordenada(x, "lon")
)

display(
    df_limpia[
        ["id_inmueble", "lat_original", "lat",
         "lon_original", "lon"]
    ]
)

```

```python

df_limpia["coordenadas_validas"] = (
    df_limpia["lat"].between(4, 13) &
    df_limpia["lon"].between(-80, -66)
)

print("Coordenadas válidas:")
display(
    df_limpia[
        ["id_inmueble", "municipio_ubicacion",
         "barrio_ubicacion", "lat", "lon",
         "coordenadas_validas"]
    ]
)
```

```python

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Configurar el servicio geográfico
geolocator = Nominatim(
    user_agent="proyecto_analisis_inmobiliario_eafit"
)

# Limitar las consultas para respetar el servicio
reverse = RateLimiter(
    geolocator.reverse,
    min_delay_seconds=1.1,
    swallow_exceptions=True
)

# Crear una nueva columna sin alterar la dirección original
df_limpia["direccion_referencial"] = df_limpia["direccion"]

df_limpia["fuente_direccion"] = np.where(
    df_limpia["direccion"].notna(),
    "Direccion original",
    "Pendiente de geocodificacion"
)

# Consultar solamente los registros sin dirección
for indice, fila in df_limpia.iterrows():

    if pd.isna(fila["direccion"]) and fila["coordenadas_validas"]:

        coordenada = (
            fila["lat"],
            fila["lon"]
        )

        ubicacion = reverse(
            coordenada,
            language="es",
            zoom=18
        )

        if ubicacion is not None:

            df_limpia.at[
                indice,
                "direccion_referencial"
            ] = ubicacion.address

            df_limpia.at[
                indice,
                "fuente_direccion"
            ] = "OpenStreetMap - referencia aproximada"

        else:

            df_limpia.at[
                indice,
                "fuente_direccion"
            ] = "Sin resultado geografico"

print("Consulta geográfica finalizada")

display(
    df_limpia[
        [
            "id_inmueble",
            "municipio_ubicacion",
            "zona_ubicacion",
            "barrio_ubicacion",
            "lat",
            "lon",
            "direccion",
            "direccion_referencial",
            "fuente_direccion"
        ]
    ]
)
```

```python

import pandas as pd
import numpy as np
import time

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Cargar nuevamente el archivo original
df = pd.read_excel("base_inmobiliaria_consolidada.xlsx")

# Normalizar coordenadas
def normalizar_coordenada(valor, tipo):

    if pd.isna(valor):
        return np.nan

    numero = float(valor)

    if tipo == "lat" and -90 <= numero <= 90:
        return numero

    if tipo == "lon" and -180 <= numero <= 180:
        return numero

    signo = -1 if numero < 0 else 1
    digitos = str(int(abs(numero)))

    if tipo == "lat":
        decimales = len(digitos) - 1
    else:
        decimales = len(digitos) - 2

    return signo * int(digitos) / (10 ** decimales)


df["lat"] = df["lat"].apply(
    lambda x: normalizar_coordenada(x, "lat")
)

df["lon"] = df["lon"].apply(
    lambda x: normalizar_coordenada(x, "lon")
)

print("Coordenadas normalizadas.")
```

```python

# Identificar inmuebles sin dirección
sin_direccion = df[df["direccion"].isna()].copy()

# Crear identificador de coordenadas
sin_direccion["clave_geo"] = list(zip(
    sin_direccion["lat"].round(6),
    sin_direccion["lon"].round(6)
))

# Obtener coordenadas únicas
puntos_unicos = sin_direccion[
    ["lat", "lon", "municipio_ubicacion",
     "barrio_ubicacion"]
].drop_duplicates(subset=["lat", "lon"])

print("Inmuebles sin dirección:", len(sin_direccion))
print("Puntos geográficos únicos:", len(puntos_unicos))

display(puntos_unicos)
```

```python

# Configurar Nominatim
geolocator = Nominatim(
    user_agent="eafit_proyecto_analisis_inmobiliario"
)

# Respetar el límite de consultas
reverse = RateLimiter(
    geolocator.reverse,
    min_delay_seconds=1.2,
    swallow_exceptions=True
)

# Diccionario para almacenar resultados
resultados_geo = {}

for _, fila in puntos_unicos.iterrows():

    lat = fila["lat"]
    lon = fila["lon"]

    clave = (round(lat, 6), round(lon, 6))

    ubicacion = reverse(
        (lat, lon),
        exactly_one=True,
        language="es",
        zoom=17,
        addressdetails=True
    )

    if ubicacion is not None:

        direccion = ubicacion.raw.get("address", {})

        resultados_geo[clave] = {
            "calle_osm": direccion.get("road"),
            "numero_osm": direccion.get("house_number"),
            "barrio_osm": (
                direccion.get("neighbourhood")
                or direccion.get("suburb")
            ),
            "municipio_osm": (
                direccion.get("city")
                or direccion.get("town")
                or direccion.get("municipality")
            ),
            "direccion_osm": ubicacion.address,
            "estado_consulta": "Resultado obtenido"
        }

    else:

        resultados_geo[clave] = {
            "calle_osm": None,
            "numero_osm": None,
            "barrio_osm": None,
            "municipio_osm": None,
            "direccion_osm": None,
            "estado_consulta": "Sin resultado"
        }

    print("Consulta finalizada:", clave)
```

```python

# Crear columnas nuevas para no alterar las originales
columnas_geo = [
    "calle_osm",
    "numero_osm",
    "barrio_osm",
    "municipio_osm",
    "direccion_osm",
    "estado_consulta"
]

for columna in columnas_geo:
    df[columna] = np.nan

# Incorporar resultados a cada inmueble sin dirección
for indice, fila in df.iterrows():

    if pd.isna(fila["direccion"]):

        clave = (
            round(fila["lat"], 6),
            round(fila["lon"], 6)
        )

        datos = resultados_geo.get(clave, {})

        for columna in columnas_geo:
            df.at[indice, columna] = datos.get(columna)


# Crear dirección aproximada
def construir_direccion(fila):

    if pd.notna(fila["direccion"]):
        return fila["direccion"]

    if pd.notna(fila["calle_osm"]):

        calle = fila["calle_osm"]
        numero = fila["numero_osm"]

        if pd.notna(numero):
            return f"{calle} #{numero}"

        return calle

    return (
        f"Referencia geográfica: "
        f"{fila['barrio_ubicacion']}, "
        f"{fila['municipio_ubicacion']}"
    )


df["direccion_aproximada"] = df.apply(
    construir_direccion,
    axis=1
)

# Identificar el origen de la dirección
df["fuente_direccion"] = np.where(
    df["direccion"].notna(),
    "Original",
    "OpenStreetMap o referencia de barrio"
)

display(
    df[[
        "id_inmueble",
        "municipio_ubicacion",
        "barrio_ubicacion",
        "lat",
        "lon",
        "direccion",
        "direccion_aproximada",
        "estado_consulta"
    ]]
)
```
